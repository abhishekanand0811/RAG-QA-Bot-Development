```python
"""
Business RAG QA Bot
"""

import os
import time
import json
import numpy as np
import pandas as pd
from typing import List, Dict

import openai
import tiktoken
from dotenv import load_dotenv

from pinecone import Pinecone, ServerlessSpec


class BusinessRAGBot:

    def __init__(self, openai_api_key: str, pinecone_api_key: str, pinecone_environment: str):
        openai.api_key = openai_api_key

        self.pinecone_client = Pinecone(api_key=pinecone_api_key)
        self.pinecone_environment = pinecone_environment
        self.index_name = "business-qa-bot"
        self.embedding_model = "text-embedding-3-small"
        self.chat_model = "gpt-4"
        self.chunk_size = 500
        self.chunk_overlap = 50
        self.tokenizer = tiktoken.get_encoding("cl100k_base")

    def setup_pinecone_index(self, dimension: int = 1536) -> bool:
        try:
            spec = ServerlessSpec(cloud="aws", region=self.pinecone_environment)

            existing_indexes = self.pinecone_client.list_indexes().names()
            if self.index_name not in existing_indexes:
                self.pinecone_client.create_index(
                    name=self.index_name,
                    dimension=dimension,
                    metric="cosine",
                    spec=spec
                )
                print(f"[INFO] Created new Pinecone index: {self.index_name}")
            else:
                print(f"[INFO] Using existing Pinecone index: {self.index_name}")

            self.index = self.pinecone_client.Index(self.index_name)
            return True

        except Exception as exc:
            print(f"[ERROR] Failed to set up Pinecone index: {exc}")
            return False

    def chunk_text(self, text: str, max_tokens: int = 500) -> List[str]:
        sentences = text.split('. ')
        chunks = []
        current_chunk = ""
        current_tokens = 0

        for sentence in sentences:
            sentence_tokens = len(self.tokenizer.encode(sentence))
            if current_tokens + sentence_tokens > max_tokens and current_chunk:
                chunks.append(current_chunk.strip())
                current_chunk = sentence + ". "
                current_tokens = sentence_tokens
            else:
                current_chunk += sentence + ". "
                current_tokens += sentence_tokens

        if current_chunk:
            chunks.append(current_chunk.strip())

        return chunks

    def generate_embeddings(self, texts: List[str]) -> List[List[float]]:
        try:
            response = openai.embeddings.create(
                model=self.embedding_model,
                input=texts
            )
            return [record.embedding for record in response.data]
        except Exception as exc:
            print(f"[ERROR] Embedding generation failed: {exc}")
            return []

    def store_documents(self, documents: List[Dict[str, str]]):
        vectors = []

        for doc_id, doc in enumerate(documents):
            chunks = self.chunk_text(doc['content'])
            embeddings = self.generate_embeddings(chunks)

            for chunk_id, (chunk, embedding) in enumerate(zip(chunks, embeddings)):
                vector_id = f"doc_{doc_id}_chunk_{chunk_id}"
                metadata = {
                    'content': chunk,
                    'title': doc['title'],
                    'category': doc['category'],
                    'doc_id': doc_id,
                    'chunk_id': chunk_id
                }
                vectors.append({
                    'id': vector_id,
                    'values': embedding,
                    'metadata': metadata
                })

        batch_size = 100
        for i in range(0, len(vectors), batch_size):
            batch = vectors[i:i + batch_size]
            self.index.upsert(vectors=batch)
            print(f"[INFO] Upserted batch {i // batch_size + 1}")

        print(f"[INFO] Stored {len(vectors)} vectors from {len(documents)} documents.")

    def retrieve_relevant_chunks(self, query: str, top_k: int = 5) -> List[Dict]:
        embedding_list = self.generate_embeddings([query])
        if not embedding_list:
            return []

        embedding = embedding_list[0]
        search_results = self.index.query(
            vector=embedding,
            top_k=top_k,
            include_metadata=True
        )

        return [
            {
                'content': match['metadata']['content'],
                'title': match['metadata']['title'],
                'category': match['metadata']['category'],
                'score': match['score']
            }
            for match in search_results['matches']
        ]

    def generate_answer(self, query: str, context_chunks: List[Dict]) -> str:

        context = "\n\n".join([
            f"Document: {chunk['title']} (Category: {chunk['category']})\n{chunk['content']}"
            for chunk in context_chunks
        ])

        prompt = (
            "You are a helpful business assistant. Use the context below to answer the user's question. "
            "If the answer is not found, state clearly that the information is not available.\n\n"
            f"Context:\n{context}\n\n"
            f"Question: {query}\n\nAnswer:"
        )

        try:
            response = openai.chat.completions.create(
                model=self.chat_model,
                messages=[
                    {"role": "system", "content": "You are a knowledgeable business assistant."},
                    {"role": "user", "content": prompt}
                ],
                max_tokens=500,
                temperature=0.1
            )
            return response.choices[0].message.content.strip()
        except Exception as exc:
            return f"[ERROR] Failed to generate answer: {exc}"

    def ask_question(self, query: str, top_k: int = 5) -> Dict:
        print(f"[QUERY] {query}")
        context_chunks = self.retrieve_relevant_chunks(query, top_k)
        answer = self.generate_answer(query, context_chunks)

        return {
            'query': query,
            'answer': answer,
            'sources': context_chunks,
            'num_sources': len(context_chunks)
        }

def create_sample_business_data():
    return [
        {
            'title': 'Company Policies - Remote Work',
            'category': 'HR Policies',
            'content': 'Employees can work from home up to 3 days per week with manager approval. Equipment and communication must be maintained.'
        },
        {
            'title': 'Employee Benefits Package',
            'category': 'HR Benefits',
            'content': 'Includes health, dental, vision insurance, 401(k) matching, paid time off, sick leave, and parental leave.'
        },
        {
            'title': 'IT Security Guidelines',
            'category': 'IT Security',
            'content': 'Use strong passwords, 2FA, company-approved tools, and report suspicious activity. Quarterly training is required.'
        }

    ]

def main():
    print("========== Business RAG QA Bot Demo ==========")

    OPENAI_API_KEY = "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"
    PINECONE_API_KEY = "pcsk_5P33UY_5Q1xccvxpn8tQLmBjKw6jZwK2N8zJiHtxaViMY53Eiy52sDviJJgy67EkYRL8uz"
    PINECONE_ENVIRONMENT = "us-east-1"

    rag_bot = BusinessRAGBot(OPENAI_API_KEY, PINECONE_API_KEY, PINECONE_ENVIRONMENT)

    print("Step 1: Setting up Pinecone index...")
    if not rag_bot.setup_pinecone_index():
        print("[FATAL] Could not set up Pinecone index. Exiting.")
        return

    print("Step 2: Loading sample documents...")
    documents = create_sample_business_data()

    print("Step 3: Storing documents in Pinecone...")
    rag_bot.store_documents(documents)

    print("Step 4: Waiting for index to sync...")
    time.sleep(10)

    print("Step 5: Running sample queries...\n" + "="*50)
    sample_questions = [
        "What are the main responsibilities for this internship/role?",
        "How are interns or new employees integrated into the team?",
        "Are there flexible working hours or part-time options for interns?",
        "How would you describe the company culture?",
        "What skills or qualities are most valued in this position?",

    ]

    for question in sample_questions:
        result = rag_bot.ask_question(question)
        print(f"\nQ: {result['query']}\nA: {result['answer']}")
        print("Sources:")
        for source in result['sources']:
            print(f" - {source['title']} (Score: {source['score']:.3f})")
        print("=" * 50)

    print("RAG QA Bot demo completed!")

if __name__ == "__main__":
    main()
```
